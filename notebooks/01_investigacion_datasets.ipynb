## 1. Importacion de librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

RAW_DATA_DIR = Path('../data/raw')
print('Carpeta de datos:', RAW_DATA_DIR.absolute())
```

## 2. Carga del dataset ISOT (principal)

ISOT es nuestro dataset principal porque ya contiene el texto completo de las noticias.

**Fuente:** https://www.uvic.ca/engineering/ece/isot/datasets/fake-news/index.php

```python
isot_true = pd.read_csv(RAW_DATA_DIR / 'True.csv')
isot_fake = pd.read_csv(RAW_DATA_DIR / 'Fake.csv')

isot_true['label'] = 1
isot_fake['label'] = 0

print(f'Noticias verdaderas: {len(isot_true):,}')
print(f'Noticias falsas:     {len(isot_fake):,}')
print(f'Total:               {len(isot_true) + len(isot_fake):,}')
```

## 3. Exploracion de columnas

```python
print('Columnas del dataset True:')
print(isot_true.columns.tolist())
print('\nPrimera fila:')
isot_true.head(1)
```

```python
print('Tipos de datos:')
print(isot_true.dtypes)
print('\nValores nulos:')
print(isot_true.isnull().sum())
```

## 4. Combinacion de datasets

```python
isot_completo = pd.concat([isot_true, isot_fake], ignore_index=True)
isot_completo = isot_completo.sample(frac=1, random_state=42).reset_index(drop=True)

print(f'Dataset combinado: {len(isot_completo):,} filas')
isot_completo.head()
```

## 5. Distribucion de clases

```python
fig, ax = plt.subplots(figsize=(8, 5))
isot_completo['label'].value_counts().plot(
    kind='bar',
    color=['#ef4444', '#10b981'],
    ax=ax
)
ax.set_xticklabels(['Falsa (0)', 'Verdadera (1)'], rotation=0)
ax.set_title('Distribucion de clases - Dataset ISOT', fontsize=14, fontweight='bold')
ax.set_ylabel('Cantidad de noticias')
plt.tight_layout()
plt.savefig('../data/processed/distribucion_clases.png', dpi=100, bbox_inches='tight')
plt.show()
```

## 6. Estadisticas de longitud de los textos

```python
isot_completo['palabras'] = isot_completo['text'].str.split().str.len()

print('Estadisticas de palabras por noticia:')
print(isot_completo['palabras'].describe())
```

```python
fig, ax = plt.subplots(figsize=(10, 5))
isot_completo[isot_completo['palabras'] < 1500]['palabras'].hist(
    bins=50, color='#2563eb', edgecolor='white'
)
ax.set_title('Distribucion de longitud de noticias (en palabras)', fontsize=14, fontweight='bold')
ax.set_xlabel('Numero de palabras')
ax.set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

## 7. Columnas relevantes identificadas

| Columna | Tipo | Usar en entrenamiento? |
